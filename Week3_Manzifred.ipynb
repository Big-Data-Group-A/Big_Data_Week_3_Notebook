## Part 1 - Making Decisions (5 marks)

### Exercise 1.1 - The Grading Ladder (3 marks)

Using if/elif/else to assign letter grades: A (>=85), B (70-84), C (60-69), D (50-59), F (<50).  
Testing with scores: 91, 76, 64, 52, 43

```python
test_scores = [91, 76, 64, 52, 43]

for score in test_scores:
    if score >= 85:
        grade = "A"
    elif score >= 70:
        grade = "B"
    elif score >= 60:
        grade = "C"
    elif score >= 50:
        grade = "D"
    else:
        grade = "F"
    print(f"Score {score} -> Grade {grade}")
```

### Exercise 1.2 - Combined Conditions (2 marks)

AUCA rule: a course is validated only if score >= 50 and attendance >= 75.  
Testing with: (76, 92), (76, 68), (43, 95)

```python
test_cases = [(76, 92), (76, 68), (43, 95)]

for score, attendance in test_cases:
    if score >= 50 and attendance >= 75:
        status = "Validated"
    elif score >= 50 and attendance < 75:
        status = "Blocked: low attendance"
    else:
        status = "Failed"
    print(f"score={score}, attendance={attendance}% -> {status}")
```

### Question 1.3 - Think About It

**Q:** What would go wrong if `elif score >= 50:` came before `elif score >= 70:`?

**Answer:** Python checks conditions from top to bottom and stops at the first one that is true. If `>= 50` comes before `>= 70`, then a score like 76 would match `>= 50` first and get grade D instead of B. The ladder has to go from the highest threshold down to the lowest, otherwise the middle grades never get reached.

## Part 2 - Loops and the Accumulator (7 marks)

```python
# score list given for part 2
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]
```

### Exercise 2.1 - Loop and Filter (2 marks)

Print only scores that are 80 or above, using a for loop.

### Exercise 2.2 - The Accumulator Pattern (3 marks)

Using one loop with no built-in functions (no sum, len, or max) to compute the total, count, average, and how many passed or failed.

```python
# start all counters at zero before the loop
total  = 0
count  = 0
passed = 0
failed = 0

for s in scores:
    total = total + s
    count = count + 1
    if s >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

average = total / count

print(f"Total:   {total}")
print(f"Count:   {count}")
print(f"Average: {average:.1f}")
print(f"Passed:  {passed}")
print(f"Failed:  {failed}")
```

```python
goal    = 250000
monthly = 15000
balance = 0
months  = 0

while balance < goal:
    balance = balance + monthly
    months  = months + 1

print(f"I will reach my goal in {months} months")
print(f"Final balance: {balance} RWF")
```

## Part 3 - Functions (6 marks)

### Exercise 3.1 - get_grade (3 marks)

Turning the grading ladder from 1.1 into a reusable function that returns the grade. Testing on 91, 76, 43.

```python
def get_grade(score):
    """Return the letter grade for a given score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    return "F"

print(get_grade(91))
print(get_grade(76))
print(get_grade(43))
```

### Exercise 3.2 - pass_rate (3 marks)

A function that takes a list of scores and returns the percentage that are 50 or above. Using an accumulator inside the function, not any built-ins.

```python
def pass_rate(score_list):
    """Return the percentage of scores that are 50 or above."""
    passed = 0
    total  = 0
    for s in score_list:
        total  = total + 1
        if s >= 50:
            passed = passed + 1
    return (passed / total) * 100

print(f"Pass rate: {pass_rate(scores):.1f}%")  # expected 90.0
```

## Part 5 - The Resilient Pipeline (12 marks)

### Exercise 5.1 - Survive the Mess (4 marks)

Going through all 40 rows and trying to convert the score to an integer. When it fails, I catch the error and note which student had the bad score. The dataset has exactly 3 broken records.

```python
total   = 0
valid   = 0
bad     = 0
bad_ids = []

for line in lines[1:]:  # skip header
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        total = total + score
        valid = valid + 1
    except ValueError:
        # score is something like N/A, absent, or empty
        bad = bad + 1
        bad_ids.append(parts[0])

average_valid = total / valid

print(f"Valid: {valid}")
print(f"Bad: {bad}")
print(f"Bad IDs: {bad_ids}")
print(f"Average: {average_valid:.1f}")
```

### Exercise 5.2 - Grades and the Top Student (4 marks)

From the valid rows, counting how many passed and failed, and finding the student with the highest score.

```python
passed     = 0
failed     = 0
best_score = -1
best_name  = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        name  = parts[1]

        if score >= 50:
            passed = passed + 1
        else:
            failed = failed + 1

        if score > best_score:
            best_score = score
            best_name  = name

    except ValueError:
        pass  # skip broken rows

print(f"Passed: {passed}")
print(f"Failed: {failed}")
print(f"Top student: {best_name}, score: {best_score}, grade: {get_grade(best_score)}")
```

### Exercise 5.3 - Report by District (4 marks)

Using two dictionaries to compute the average score per district and see which one performed the best.

```python
sums   = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    d = parts[5]  # district column
    try:
        s = int(parts[7])
        sums[d]   = sums.get(d, 0)   + s
        counts[d] = counts.get(d, 0) + 1
    except ValueError:
        pass

best_district     = ""
best_district_avg = -1

for district in sorted(sums):
    avg = sums[district] / counts[district]
    print(f"{district}: {avg:.2f}")
    if avg > best_district_avg:
        best_district_avg = avg
        best_district     = district

print(f"\nBest district: {best_district} with average {best_district_avg:.2f}")
```
