## Grading ladder

```python
test_scores = [91, 76, 64, 52, 43]

for score in test_scores:
    if score >= 85:
        grade = "A"
    elif score >= 70:
        grade = "B"
    elif score >= 60:
        grade = "C"
    elif score >= 50:
        grade = "D"
    else:
        grade = "F"
    print(score, "->", grade)
```

## Course validation

```python
test_cases = [(76, 92), (76, 68), (43, 95)]

for score, attendance in test_cases:
    if score >= 50 and attendance >= 75:
        status = "Validated"
    elif score >= 50 and attendance < 75:
        status = "Blocked: low attendance"
    else:
        status = "Failed"
    print(f"score={score}, att={attendance}% => {status}")
```

If elif score >= 50 came before elif score >= 70, a score like 76 would match the 50 condition first and get grade D instead of B. Python stops at the first true condition so the higher thresholds have to come first.

## Loops

```python
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

for score in scores:
    if score >= 80:
        print("Top score:", score)
```

```python
total = 0
count = 0
passed = 0
failed = 0

for s in scores:
    total = total + s
    count = count + 1
    if s >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

average = total / count
print("total:", total)
print("count:", count)
print("average:", round(average, 1))
print("passed:", passed)
print("failed:", failed)
```

```python
balance = 0
months = 0

while balance < 250000:
    balance = balance + 15000
    months = months + 1

print("months needed:", months)
print("balance:", balance)
```

## Functions

```python
def get_grade(score):
    """returns letter grade for a score"""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    return "F"

print(get_grade(91))
print(get_grade(76))
print(get_grade(43))
```

```python
def pass_rate(score_list):
    """returns percentage of scores that are 50 or above"""
    passed = 0
    total = 0
    for s in score_list:
        total = total + 1
        if s >= 50:
            passed = passed + 1
    return (passed / total) * 100

print(pass_rate(scores))
```

## Pipeline

```python
total = 0
valid = 0
bad = 0
bad_ids = []

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        total = total + score
        valid = valid + 1
    except ValueError:
        bad = bad + 1
        bad_ids.append(parts[0])

average_valid = total / valid

print("valid:", valid)
print("bad:", bad)
print("bad ids:", bad_ids)
print("average:", round(average_valid, 1))
```

```python
passed = 0
failed = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        name = parts[1]
        if score >= 50:
            passed = passed + 1
        else:
            failed = failed + 1
        if score > best_score:
            best_score = score
            best_name = name
    except ValueError:
        pass

print("passed:", passed)
print("failed:", failed)
print("top student:", best_name)
print("top score:", best_score)
print("top grade:", get_grade(best_score))
```

```python
sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    d = parts[5]
    try:
        s = int(parts[7])
        sums[d] = sums.get(d, 0) + s
        counts[d] = counts.get(d, 0) + 1
    except ValueError:
        pass

best_district = ""
best_avg = -1

for d in sorted(sums):
    avg = sums[d] / counts[d]
    print(d, "->", round(avg, 2))
    if avg > best_avg:
        best_avg = avg
        best_district = d

print("best district:", best_district, round(best_avg, 2))
```
