## Part 1 · Making Decisions (5 marks)

### Exercise 1.1 — The Grading Ladder (3 marks)
Given a variable `score`, print the letter grade using an `if/elif/else` ladder:

| Grade | Threshold |
|---|---|
| A | ≥ 85 |
| B | 70 – 84 |
| C | 60 – 69 |
| D | 50 – 59 |
| F | < 50 |

Test with: **91, 76, 64, 52, 43**

````python
# ── Exercise 1.1 — The Grading Ladder ──────────────────────
print("Exercise 1.1 — Grading Ladder")
print("-" * 32)

test_scores = [91, 76, 64, 52, 43]

for score in test_scores:
    if score >= 85:
        grade = "A"
    elif score >= 70:
        grade = "B"
    elif score >= 60:
        grade = "C"
    elif score >= 50:
        grade = "D"
    else:
        grade = "F"
    print(f"  Score {score:>3}  ->  Grade {grade}")
````

### Exercise 1.2 — Combined Conditions (2 marks)
**AUCA rule:** a course is validated only if `score >= 50` **and** `attendance >= 75`.

Print one of: `"Validated"` · `"Blocked: low attendance"` · `"Failed"`

Test with: **(76, 92)**, **(76, 68)**, **(43, 95)**

````python
# ── Exercise 1.2 — Combined Conditions ─────────────────────
print("Exercise 1.2 — Combined Conditions")
print("-" * 42)

test_cases = [(76, 92), (76, 68), (43, 95)]

for score, attendance in test_cases:
    if score >= 50 and attendance >= 75:
        status = "Validated"
    elif score >= 50 and attendance < 75:
        status = "Blocked: low attendance"
    else:
        status = "Failed"
    print(f"  Score={score}, Attendance={attendance}%  ->  {status}")
````

### Question 1.3 — Think About It *(included in Part 1 marks)*

**Q:** What would go wrong if `elif score >= 50:` came **before** `elif score >= 70:`?

**Answer:** Python evaluates `elif` conditions top to bottom and stops at the **first True** match. If `elif score >= 50:` appeared before `elif score >= 70:`, then any score between 50 and 84 — which should earn a **B** or **C** — would be incorrectly assigned grade **D**, because values like 76 and 64 both satisfy `>= 50` and Python would never reach the higher thresholds. The ladder must always go from the **largest to the smallest** threshold.

### Exercise 2.2 — The Accumulator Pattern (3 marks)
Using **one loop** and **NO built-ins** (`sum`, `len`, `max` forbidden):
Compute **(a)** total · **(b)** count · **(c)** average · **(d)** passed (≥ 50) and failed counts.

````python
# ── Exercise 2.2 — Accumulator Pattern ─────────────────────
print("Exercise 2.2 — Accumulator Pattern")
print("-" * 36)

# Step 1: Initialize BEFORE the loop
total  = 0
count  = 0
passed = 0
failed = 0

# Step 2: Update INSIDE the loop
for s in scores:
    total = total + s
    count = count + 1
    if s >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

# Step 3: Report AFTER the loop
average = total / count

print(f"  (a) Total   : {total}")
print(f"  (b) Count   : {count}")
print(f"  (c) Average : {average:.1f}")
print(f"  (d) Passed  : {passed}")
print(f"      Failed  : {failed}")
````

### Exercise 2.3 — while Loop (2 marks)
You save **15,000 RWF** per month toward a **250,000 RWF** laptop.
Use a `while` loop to compute how many months you need.

````python
# ── Exercise 2.3 — while Loop ──────────────────────────────
print("Exercise 2.3 — Laptop Savings Goal")
print("-" * 36)

goal    = 250_000   # target in RWF
monthly = 15_000    # saved per month
balance = 0
months  = 0

# balance grows every iteration — no infinite loop risk
while balance < goal:
    balance = balance + monthly
    months  = months + 1

print(f"  Monthly saving : {monthly:,} RWF")
print(f"  Goal           : {goal:,} RWF")
print(f"  Months needed  : {months}")
print(f"  Final balance  : {balance:,} RWF")
````

### Exercise 3.1 — get_grade (3 marks)
Turn the 1.1 ladder into `get_grade(score)` that **returns** the letter grade.  
Include a **one-line docstring**. Test on **91, 76, 43**.

````python
# ── Exercise 3.1 — get_grade() ─────────────────────────────
def get_grade(score):
    """Return the letter grade A-F for a given numeric score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    return "F"

print("Exercise 3.1 — get_grade()")
print("-" * 30)
for s in [91, 76, 43]:
    print(f"  get_grade({s}) -> {get_grade(s)}")
````

### Exercise 3.2 — pass_rate (3 marks)
Write `pass_rate(score_list)` that returns the **percentage** of scores ≥ 50 using an **accumulator** inside the function.  
Expected result on the Part 2 list: **90.0**

````python
# ── Exercise 3.2 — pass_rate() ─────────────────────────────
def pass_rate(score_list):
    """Return the percentage of scores >= 50 in score_list."""
    passed = 0
    total  = 0
    for s in score_list:
        total  = total + 1
        if s >= 50:
            passed = passed + 1
    return (passed / total) * 100

print("Exercise 3.2 — pass_rate()")
print("-" * 36)
result = pass_rate(scores)
print(f"  pass_rate(scores) = {result:.1f}%   (expected: 90.0%)")
````

## Part 5 · The Resilient Pipeline (12 marks)

### Exercise 5.1 — Survive the Mess (4 marks)
Loop over all rows. `try` to convert `parts[7]` to `int`:  
- **Success** → add to `total`, increment `valid`  
- **ValueError** → increment `bad`, append `student_id` to `bad_ids`

Print valid, bad, bad_ids, and the average of valid scores (1 decimal).

> ✅ **Checkpoint: exactly 3 broken records.**

````python
# ── Exercise 5.1 — Survive the Mess ────────────────────────
print("Exercise 5.1 — Resilient Pipeline")
print("-" * 42)

total   = 0
valid   = 0
bad     = 0
bad_ids = []

for line in lines[1:]:                   # skip header
    parts = line.strip().split(",")
    try:
        score = int(parts[7])            # will raise ValueError on "N/A", "", "absent"
        total = total + score
        valid = valid + 1
    except ValueError:
        bad = bad + 1
        bad_ids.append(parts[0])         # log the broken student_id

average_valid = total / valid

print(f"  Valid records   : {valid}")
print(f"  Bad  records    : {bad}")
print(f"  Bad student IDs : {bad_ids}")
print(f"  Average score   : {average_valid:.1f}")
print()
if bad == 3:
    print("  Checkpoint PASSED — exactly 3 broken records found!")
else:
    print(f"  Unexpected result — found {bad} broken records.")
````

### Exercise 5.2 — Grades & the Top Student (4 marks)
Among **valid rows only**:
- **(a)** count passed (≥ 50) and failed  
- **(b)** track the highest score and the student's name with two accumulators  

Print both, plus the top student's letter grade via `get_grade`.

````python
# ── Exercise 5.2 — Grades & Top Student ────────────────────
print("Exercise 5.2 — Grades & Top Student")
print("-" * 42)

passed     = 0
failed     = 0
best_score = -1    # accumulator: highest score seen so far
best_name  = ""    # accumulator: that student's name

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        name  = parts[1]

        # (a) pass / fail tally
        if score >= 50:
            passed = passed + 1
        else:
            failed = failed + 1

        # (b) track top student
        if score > best_score:
            best_score = score
            best_name  = name

    except ValueError:
        pass   # broken rows already identified in 5.1

print(f"  (a) Passed      : {passed}")
print(f"      Failed      : {failed}")
print(f"  (b) Top student : {best_name}")
print(f"      Top score   : {best_score}")
print(f"      Top grade   : {get_grade(best_score)}")
````

### Exercise 5.3 — Stretch: Report by District (4 marks)
Which district has the highest average score? Use **two dictionary accumulators** — `sums` and `counts`.  
There are **8 districts, 5 students each**.

````python
# ── Exercise 5.3 — Report by District ──────────────────────
print("Exercise 5.3 — Average Score by District")
print("-" * 44)

sums   = {}   # district -> running total
counts = {}   # district -> student count

for line in lines[1:]:
    parts = line.strip().split(",")
    d = parts[5]                              # index 5 = district
    try:
        s = int(parts[7])
        sums[d]   = sums.get(d, 0)   + s
        counts[d] = counts.get(d, 0) + 1
    except ValueError:
        pass

best_district     = ""
best_district_avg = -1

print(f"  {'District':<16} {'Students':>8}  {'Avg Score':>10}")
print(f"  {'-'*16} {'-'*8}  {'-'*10}")

for district in sorted(sums):
    avg = sums[district] / counts[district]
    print(f"  {district:<16} {counts[district]:>8}  {avg:>10.2f}")
    if avg > best_district_avg:
        best_district_avg = avg
        best_district     = district

print()
print(f"  Leading district: {best_district}  (avg: {best_district_avg:.2f})")
````

## Part 6 · Reflection *(required, checked for completion)*

**Question 1:** Your pipeline skipped 3 broken records out of 40 (7.5%). At what percentage would you **stop trusting** the dataset — and what would you do **instead of skipping**?

**Answer:**  
At 7.5% (3 out of 40 records), skipping is acceptable: the remaining 37 valid records are enough to draw meaningful conclusions, and the error rate is low enough not to bias averages significantly. However, I would stop trusting the dataset and change my approach once the bad-record rate crosses **15%**. At that point, the students with missing scores might share a pattern — for example, they could all be from one district or program — which means silently skipping them would introduce **selection bias** and distort the results.

Instead of skipping, I would:
1. **Investigate the root cause** — contact the data source (registrar or exam system) to understand why values are missing.
2. **Impute cautiously** — replace missing scores with the district or program average, and flag those rows clearly so any reader knows they were estimated.
3. **Report transparently** — always state the missing-data rate in any output or report. Silent data loss is the analyst's deadliest sin.
