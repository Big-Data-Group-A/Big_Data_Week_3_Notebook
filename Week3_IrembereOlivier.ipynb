```python
#Exercise 1.1 — The grading ladder
def get_grade(score):
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"
# Test scores
scores = [91, 76, 64, 52, 43]

for score in scores:
    print(f"Score: {score} → Grade: {get_grade(score)}")
```

```python
#Exercise 1.2 — Combined conditions
def check_validation(score, attendance):
    if score >= 50 and attendance >= 75:
        return "Validated"
    elif score >= 50 and attendance < 75:
        return "Blocked: low attendance"
    else:
        return "Failed"


# Test cases
students = [(76, 92), (76, 68), (43, 95)]

for score, attendance in students:
    print(f"Score: {score}, Attendance: {attendance}% → {check_validation(score, attendance)}")
```

Part 2 · Loops & the Accumulator


```python
#Exercise 2.1 — Loop + filter
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

for score in scores:
    if score >= 80:
        print(f"Top score: {score}")
```

```python
#Exercise 2.2 — The accumulator pattern
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

# Initialize
total = 0
count = 0
passed = 0
failed = 0

# Update inside the loop
for score in scores:
    total = total + score
    count = count + 1

    if score >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

# Report after the loop
average = total / count

print("Total:", total)
print("Count:", count)
print("Average:", average)
print("Passed:", passed)
print("Failed:", failed)
```

```python
#Exercise 2.3 — while loop
monthly_saving = 15000
laptop_cost = 250000

saved = 0
months = 0

while saved < laptop_cost:
    saved = saved + monthly_saving
    months = months + 1

print("Months needed:", months)
print("Total saved:", saved, "RWF")
```

Part 3 · **Functions**

```python
#Exercise 3.1 — get_grade
def get_grade(score):
    """Return the letter grade for a given score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"


# Test the function
print(get_grade(91))
print(get_grade(76))
print(get_grade(43))
```

```python
#Exercise 3.2 — pass_rate

def pass_rate(score_list):
    """Return the percentage of scores that are 50 or higher."""
    passed = 0
    count = 0

    for score in score_list:
        count = count + 1

        if score >= 50:
            passed = passed + 1

    return (passed / count) * 100


# Test the function
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

print(pass_rate(scores))
```

Part 5 · The Resilient Pipeline

```python
# Exercise 5.1 — Survive the mess

total = 0
valid = 0
bad = 0
bad_ids = []

for line in lines[1:]:
    parts = line.strip().split(",")

    student_id = parts[0]

    try:
        score = int(parts[7])

        total = total + score
        valid = valid + 1

    except ValueError:
        bad = bad + 1
        bad_ids.append(student_id)

average = total / valid

print("Valid:", valid)
print("Bad:", bad)
print("Bad IDs:", bad_ids)
print("Average:", round(average, 1))
```

```python
#Exercise 5.2 — Grades & the top student
passed = 0
failed = 0

best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")

    student_id = parts[0]
    name = parts[1]

    try:
        score = int(parts[7])

        # Count passed and failed
        if score >= 50:
            passed = passed + 1
        else:
            failed = failed + 1

        # Track the highest score
        if score > best_score:
            best_score = score
            best_name = name

    except ValueError:
        # Ignore broken scores
        continue

print("Passed:", passed)
print("Failed:", failed)
print("Best score:", best_score)
print("Best student:", best_name)
print("Top student's grade:", get_grade(best_score))
```

```python
#Exercise 5.3 — Stretch: report by district
sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")

    district = parts[5]

    try:
        score = int(parts[7])

        sums[district] = sums.get(district, 0) + score
        counts[district] = counts.get(district, 0) + 1

    except ValueError:
        continue
# Print every district's average
highest_district = ""
highest_average = -1

for district in sums:
    average = sums[district] / counts[district]

    print(district, "→", round(average, 1))

    if average > highest_average:
        highest_average = average
        highest_district = district

print("District with highest average:", highest_district)
print("Highest average:", round(highest_average, 1))
```
