## Part 1 · Making Decisions

###  1.1 — The grading ladder 

```python
score= 43;      # Tested it by changing score to each of 91, 76, 64, 52, 43
if score >=85:
    print("A");
elif score >= 70:
    print("B");
elif score >= 60:
    print("C");
elif score >= 50:
    print("D");
else:
    print("F");
```

### 1.2 — Combined conditions

```python
score =76
attendance = 92
if score >= 50 and attendance >= 75:
    print("Validated");
elif score >= 50:
    print ("Blocked: low attendance");
else:
    print("Failed");
```

### 1.3 — Think about it

Since Python stops at the first true condition it would print D because 91 is >= 50 thus, the broader conditions must come after narrower ones, not before. 

## Part 2 · Loops & the Accumulator

### 2.1 — Loop + filter

```python
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

for s in scores:
    if s >= 80:
        print(f"Top score: {s}");
```

###  2.2 — The accumulator pattern

```python
total = 0
count = 0
passed = 0
failed = 0

for s in scores:
    total = total + s
    count = count + 1
    if s >= 50:
        passed = passed + 1
    else:
        failed = failed + 1
        
average = total/count

print(f"Total: {total}")
print(f"Count: {count}")
print(f"Average: {average}")
print(f"Passed: {passed}, Failed: {failed}")
```

### 2.3 — While loop

```python
saved = 0
months = 0
monthly_saving = 15000
target = 250000

while saved < target:
    saved += monthly_saving
    months += 1

print(f"Months needed: {months}")
```

## Part 3 · Functions

### 3.1 — get_grade

```python
def get_grade (score):
    """"Return the letter grade for a given numeric score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

print(get_grade(91))
print(get_grade(76))
print(get_grade(43))
```

### 3.2 — pass_rate

```python
def pass_rate(score_list):
    """Return the percentage of scores in score_list that are >= 50."""
    count = 0
    passed = 0
    for s in score_list:
        count += 1
        if s >= 50:
            passed += 1
    return (passed / count) * 100

print(pass_rate(scores))
```
