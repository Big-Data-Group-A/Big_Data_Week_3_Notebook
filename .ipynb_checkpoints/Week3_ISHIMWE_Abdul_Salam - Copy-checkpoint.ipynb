## Part 1 · Making Decisions

###  1.1 — The grading ladder 

```python
score= 43;      # Tested it by changing score to each of 91, 76, 64, 52, 43
if score >=85:
    print("A");
elif score >= 70:
    print("B");
elif score >= 60:
    print("C");
elif score >= 50:
    print("D");
else:
    print("F");
```

### 1.2 — Combined conditions

```python
score =76
attendance = 92
if score >= 50 and attendance >= 75:
    print("Validated");
elif score >= 50:
    print ("Blocked: low attendance");
else:
    print("Failed");
```

### 1.3 — Think about it

Since Python stops at the first true condition it would print D because 91 is >= 50 thus, the broader conditions must come after narrower ones, not before. 

## Part 2 · Loops & the Accumulator

### 2.1 — Loop + filter

```python
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

for s in scores:
    if s >= 80:
        print(f"Top score: {s}");
```

###  2.2 — The accumulator pattern

```python
total = 0
count = 0
passed = 0
failed = 0

for s in scores:
    total = total + s
    count = count + 1
    if s >= 50:
        passed = passed + 1
    else:
        failed = failed + 1
        
average = total/count

print(f"Total: {total}")
print(f"Count: {count}")
print(f"Average: {average}")
print(f"Passed: {passed}, Failed: {failed}")
```

### 2.3 — While loop

```python
saved = 0
months = 0
monthly_saving = 15000
target = 250000

while saved < target:
    saved += monthly_saving
    months += 1

print(f"Months needed: {months}")
```

## Part 3 · Functions

### 3.1 — get_grade

```python
def get_grade (score):
    """"Return the letter grade for a given numeric score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

print(get_grade(91))
print(get_grade(76))
print(get_grade(43))
```

### 3.2 — pass_rate

```python
def pass_rate(score_list):
    """Return the percentage of scores in score_list that are >= 50."""
    count = 0
    passed = 0
    for s in score_list:
        count += 1
        if s >= 50:
            passed += 1
    return (passed / count) * 100

print(pass_rate(scores))
```

## Part 4 · Reading the File

### 4.1 — First contact

```python
with open("week3_students.csv") as f:
    lines = f.readlines()
print(len(lines))
print(lines[0])
print(lines[1])
      
```

### 4.2 — extract a column

```python
names = []

for line in lines[1:]:
    parts = line.split(",")
    names.append(parts[1])

print(len(names))
print(names[:5])
```

## Part 5 · The Resilient Pipeline

### 5.1 — Survive the mess 

```python
total = 0
valid = 0
bad = 0
bad_ids = []

for line in lines[1:]:
    parts = line.split(",")
    student_id = parts[0]
    try:
        score = int(parts[7])
        total += score
        valid += 1
    except ValueError:
        bad += 1
        bad_ids.append(student_id)

average = total / valid

print(f"Valid: {valid}")
print(f"Bad: {bad}")
print(f"Bad IDs: {bad_ids}")
print(f"Average of valid scores: {round(average, 1)}")
```

### 5.2 — Grades & the top student

```python
passed = 0
failed = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.split(",")
    try:
        score = int(parts[7])
    except ValueError:
        continue   # skip this row entirely, move to the next line

    if score >= 50:
        passed += 1
    else:
        failed += 1

    if score > best_score:
        best_score = score
        best_name = parts[1]

print(f"Passed: {passed}, Failed: {failed}")
print(f"Top student: {best_name} with score {best_score}")
print(f"Top student's grade: {get_grade(best_score)}")
```

###  5.3 — Stretch: report by district

```python
sums = {}
counts = {}

for line in lines[1:]:
    parts = line.split(",")
    try:
        score = int(parts[7])
    except ValueError:
        continue

    d = parts[5]  
    sums[d] = sums.get(d, 0) + score
    counts[d] = counts.get(d, 0) + 1

for d in sums:
    avg = sums[d] / counts[d]
    print(f"{d}: {round(avg, 1)}")
```

### Bonus (+3) — Your first output file

```python
with open("week3_report.txt", "w") as f:
    f.write("--- Abdul Salam's Report ---\n")
    f.write(f"Student count: {valid + bad}\n")
    f.write(f"Valid: {valid}, Bad: {bad}\n")
    f.write(f"Average score: {round(average, 1)}\n")
    f.write(f"Top student: {best_name} ({best_score})\n")
    f.write("--- End of Report ---")

with open("week3_report.txt") as f:
    print(f.read())
```

## Part 6 · Reflection

1. I would stop trusting the dataset if more than 10% of records were broken. At that point, skipping them would bias results too much. Instead of skipping, I would:

-Try to recover the values (e.g., check for typos, decimal points, or "N/A" markers).
-Flag the records for manual review rather than silently dropping them.
-Report the missing rate alongside any analysis so readers know the limitation.

2. The try/except ValueError pattern was surprising ,it lets the program keep running even when data is broken, instead of crashing. It's a simple but powerful way to build resilient pipelines.
