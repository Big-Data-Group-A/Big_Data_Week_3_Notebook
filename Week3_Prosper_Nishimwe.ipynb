### 1.1 Grading Ladder

```python
scores = [91, 76, 64, 52, 43]

for score in scores:
    if score >= 85:
        print(score, "A")
    elif score >= 70:
        print(score, "B")
    elif score >= 60:
        print(score, "C")
    elif score >= 50:
        print(score, "D")
    else:
        print(score, "F")
```

### 1.2 Combined Conditions

```python
tests = [(76, 92), (76, 68), (43, 95)]

for score, attendance in tests:
    if score >= 50 and attendance >= 75:
        print("Validated")
    elif score >= 50 and attendance < 75:
        print("Blocked: low attendance")
    else:
        print("Failed")
```

### 1.3 Why does the order of elif matter?

If `elif score >= 50` comes before `elif score >= 70`, scores of 70 or higher will be classified as D because the first true condition stops Python from checking the later conditions.

## Part 2 — Loops & Accumulator

```python
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

for score in scores:
    if score >= 80:
        print("Top score:", score)
```

### 2.2 Accumulator

```python
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

total = 0
count = 0
passed = 0
failed = 0

for score in scores:
    total += score
    count += 1

    if score >= 50:
        passed += 1
    else:
        failed += 1

average = total / count

print("Total:", total)
print("Count:", count)
print("Average:", average)
print("Passed:", passed)
print("Failed:", failed)
```

### 2.3 While Loop

```python
saved = 0
months = 0

while saved < 250000:
    saved += 15000
    months += 1

print("Months needed:", months)
print("Total saved:", saved)
```

## Part 3 — Functions

```python
def get_grade(score):
    """Return the letter grade for a score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"


print(get_grade(91))
print(get_grade(76))
print(get_grade(43))
```

```python
def pass_rate(score_list):
    passed = 0

    for score in score_list:
        if score >= 50:
            passed += 1

    return (passed / len(score_list)) * 100


scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

print("Pass rate:", pass_rate(scores), "%")
```
