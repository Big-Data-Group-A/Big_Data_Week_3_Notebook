### 1.1 Grading Ladder

```python
scores = [91, 76, 64, 52, 43]

for score in scores:
    if score >= 85:
        print(score, "A")
    elif score >= 70:
        print(score, "B")
    elif score >= 60:
        print(score, "C")
    elif score >= 50:
        print(score, "D")
    else:
        print(score, "F")
```

### 1.2 Combined Conditions

```python
tests = [(76, 92), (76, 68), (43, 95)]

for score, attendance in tests:
    if score >= 50 and attendance >= 75:
        print("Validated")
    elif score >= 50 and attendance < 75:
        print("Blocked: low attendance")
    else:
        print("Failed")
```

### 1.3 Why does the order of elif matter?

If `elif score >= 50` comes before `elif score >= 70`, scores of 70 or higher will be classified as D because the first true condition stops Python from checking the later conditions.

### 2.2 Accumulator

```python
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

total = 0
count = 0
passed = 0
failed = 0

for score in scores:
    total += score
    count += 1

    if score >= 50:
        passed += 1
    else:
        failed += 1

average = total / count

print("Total:", total)
print("Count:", count)
print("Average:", average)
print("Passed:", passed)
print("Failed:", failed)
```

### 2.3 While Loop

```python
saved = 0
months = 0

while saved < 250000:
    saved += 15000
    months += 1

print("Months needed:", months)
print("Total saved:", saved)
```

## Part 3 — Functions

```python
def get_grade(score):
    """Return the letter grade for a score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"


print(get_grade(91))
print(get_grade(76))
print(get_grade(43))
```

```python
def pass_rate(score_list):
    passed = 0

    for score in score_list:
        if score >= 50:
            passed += 1

    return (passed / len(score_list)) * 100


scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

print("Pass rate:", pass_rate(scores), "%")
```

## Part 5 — Resilient Pipeline

```python
valid_scores = []
bad_ids = []

with open(r"C:\Users\USER\Big_Data_Week_3_Notebook\week3_students.csv", "r") as f:
    lines = f.readlines()

for line in lines[1:]:
    parts = line.strip().split(",")
    
    student_id = parts[0]
    name = parts[1]
    score = parts[7]
    
    try:
        score = int(score)
        valid_scores.append(score)
    except ValueError:
        bad_ids.append(student_id)

print("Valid scores:", len(valid_scores))
print("Broken records:", len(bad_ids))
print("Broken student IDs:", bad_ids)

average = sum(valid_scores) / len(valid_scores)
print("Average valid score:", round(average, 1))
```

```python
passed = 0
failed = 0
highest_score = -1
top_student = ""
top_grade = ""

for line in lines[1:]:
    parts = line.strip().split(",")

    student_id = parts[0]
    name = parts[1]
    score_text = parts[7]

    try:
        score = int(score_text)

        if score >= 50:
            passed += 1
        else:
            failed += 1

        if score > highest_score:
            highest_score = score
            top_student = name

        if score >= 80:
            grade = "A"
        elif score >= 70:
            grade = "B"
        elif score >= 60:
            grade = "C"
        elif score >= 50:
            grade = "D"
        else:
            grade = "F"

        if score == highest_score:
            top_grade = grade

    except ValueError:
        continue

print("Passed:", passed)
print("Failed:", failed)
print("Highest score:", highest_score)
print("Top student:", top_student)
print("Top grade:", top_grade)
```
