```python
#EXERCISES 3.2

def pass_rate(score_list):
    
    passed = 0
    count = 0
    for score in score_list:
        count = count + 1
        if score >= 50:
            passed = passed + 1
    return (passed / count) * 100

scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]
print(pass_rate(scores))
```

```python
#PART 5

#EXERCISES 5.1

total = 0
count = 0
bad = 0
bad_ids = []

with open("week3_students.csv") as f:
    lines = f.readlines()

for line in lines[1:]:                  
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        total = total + score
        count = count + 1
    except ValueError:
        bad = bad + 1
        bad_ids.append(parts[0])

print(f"Valid: {count}, Bad: {bad}")
print(f"Bad IDs: {bad_ids}")
if count > 0:
    print(f"Average: {total/count:.1f}")
else:
    print("No valid scores")
```

```python
#EXERCISE 5.2

def get_grade(score):
    """Convert a score to a letter grade."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

total = 0
count = 0
bad = 0
passed = 0
failed = 0
best_score = -1
best_name = ""

with open("week3_students.csv") as f:
    lines = f.readlines()

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        total = total + score
        count = count + 1
        if score >= 50:
            passed = passed + 1
        else:
            failed = failed + 1
        if score > best_score:
            best_score = score
            best_name = parts[1]
    except ValueError:
        bad = bad + 1

print(f"Passed: {passed}, Failed: {failed}")
print(f"Top student: {best_name} with {best_score}")
print(f"Top grade: {get_grade(best_score)}")
```

```python
#EXERCISE 5.3

sums = {}
counts = {}

with open("week3_students.csv") as f:
    lines = f.readlines()

for line in lines[1:]:
    parts = line.strip().split(",")
    d = parts[5]                        
    try:
        s = int(parts[7])              
        sums[d] = sums.get(d, 0) + s
        counts[d] = counts.get(d, 0) + 1
    except ValueError:
        pass                            


best_district = ""
best_avg = -1

for d in sums:
    avg = sums[d] / counts[d]
    print(f"{d}: {avg:.1f}")
    if avg > best_avg:
        best_avg = avg
        best_district = d

print(f"\nTop district: {best_district} with average {best_avg:.1f}")
```
