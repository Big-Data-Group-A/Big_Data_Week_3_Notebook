```python
#PART 3
#3.1

def get_grade(score):
    # score = 76
    if score >= 85:
        return "A"
    elif score > 70 and score < 84:
        return "B"
    elif score > 60 and score < 69:
        return "C"
    elif score > 50 and score < 59:
        return "D"
    else:
        return "F"

print(get_grade(91))  
print(get_grade(76))
print(get_grade(43))
```

```python
#3.2

scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]
def pass_rate(score_list):
    count = 0
    result = 0
    for score in scores:
        if score >= 50:
            count += 1
    result = count / len(score_list) * 100
    return result

pass_rate(scores)
```

```python
#PART 4
#4.1

with open("week3_students.csv", "r") as f:
    lines = f.readlines()
    print (len(lines))#number of lines
    print (lines[0]) #header line
    print (lines[1]) #first student's line
    #It is 41 because of the title of columns of the table
```

```python
#4.2

names=[]
for line in lines[1:]:
    parts = line.strip().split(",")
    names.append(parts[1])
print (len(names))
print (names[:5])
```

```python
#PART 5
#5.1

total = 0
valid = 0
bad = 0
bad_ids = []

for line in lines[1:]:
    parts = line.strip().split(",")
    student_id = parts[0]
    try:
        score = int(parts[7])
        total += score
        valid += 1
    except ValueError:
        bad += 1
        bad_ids.append(student_id)

average = total / valid 

print ("Valid : ",valid," bad : ",bad)
print ("Bad IDs : ", bad_ids)
```

```python
#5.2

passed = 0
failed = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        name = parts[1]

        if score >= 50:
            passed += 1
        else:
            failed += 1

        if score > best_score:
            best_score = score
            best_name = name
    except ValueError:
        continue
best_grade = get_grade(best_score)
print("Best score = ",best_score," with names of ",best_name)
print("Top student's grade is ", best_grade)

```

```python
#5.3

sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        s = int(parts[7])
        d = parts[5]
        sums[d] = sums.get(d,0) + s
        counts[d] = counts.get(d,0) + 1
    except ValueError:
        continue
highest_avg = -1
best_district = ""

for d in sums:
    avg = sums[d] / counts[d]
    print ("District ",d," , Average ",avg)
    if avg > highest_avg:
        highest_avg = avg
        best_district = d
print ("District with highest average score is ",best_district," with average of ",highest_avg)
```

```python
#Bonus 

report = f""" 
Students Count: {len(lines) - 1}
Valid Records: {valid}
Bad Records: {bad}
Average Score: {avg:.1f}
Top Student: {best_name} ({best_score})
"""

with open("week3_report.txt", "w") as f:
    f.write(report)

with open("week3_report.txt", "r") as f:
    content = f.read()
    print(content)
```
