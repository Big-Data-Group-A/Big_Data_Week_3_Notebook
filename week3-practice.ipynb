```python
# the grading ladder (Lab Ex 1.1):
def show_grade(score):
    if score >= 85:
        grade = "A"
    elif score >= 70:
        grade = "B"
    elif score >= 60:
        grade = "C"
    elif score >= 50:
        grade = "D"
    else:
        grade = "F"
    print(f"Score {score} -> Grade {grade}")

for test in [91, 76, 64, 52, 43]:
    show_grade(test)              

    #If elif score >= 50 came before elif score >= 70, then any score of 70 or above would incorrectly stop at the 50-threshold branch and get grade "D" instead of "B" or "A" — because Python takes the first true condition and never checks the rest.
```

```python
#Exercise 1.2 — combined conditions:
def check_validation(score, attendance):
    if score >= 50 and attendance >= 75:
        print("Validated")
    elif score >= 50 and attendance < 75:
        print("Blocked: low attendance")
    else:
        print("Failed")

check_validation(76, 92)  # Validated
check_validation(76, 68)  # Blocked: low attendance
check_validation(43, 95)  # Failed
```

```python
#Exercise 2.2 — no built-ins allowed:
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

total = 0
count = 0
passed = 0
failed = 0

for s in scores:
    total = total + s
    count = count + 1
    if s >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

average = total / count
print(f"Total: {total}")
print(f"Count: {count}")
print(f"Average: {average:.1f}")
print(f"Passed: {passed}, Failed: {failed}")
```

```python
#Exercise 2.3 — while loop (savings): while is for :keep going until a condition becomes false — you don't know the number of rounds in advance, unlike a for loop over a fixed list.
balance = 0
months = 0
goal = 250000
monthly_saving = 15000

while balance < goal:
    balance = balance + monthly_saving
    months = months + 1

print(f"You need {months} months to reach {goal} RWF")
```

```python
#Exercise 3.1 — get_grade as a function:
def get_grade(score):
    """Convert a numeric score into a letter grade."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    return "F"

print(get_grade(91))  # A
print(get_grade(76))  # B
print(get_grade(43))  # F
```

```python
#Exercise 3.2 — pass_rate function (accumulator inside a function):
def pass_rate(score_list):
    """Return the percentage of scores that are >= 50."""
    total = 0
    passed = 0
    for s in score_list:
        total = total + 1
        if s >= 50:
            passed = passed + 1
    return (passed / total) * 100

scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]
print(pass_rate(scores))  # 90.0
```

```python
#Exercise 5.1 — survive the mess:
total = 0
valid = 0
bad = 0
bad_ids = []

with open("week3_students.csv") as f:
    lines = f.readlines()

for line in lines[1:]:
    parts = line.strip().split(",")
    student_id = parts[0]
    try:
        score = int(parts[7])
        total = total + score
        valid = valid + 1
    except ValueError:
        bad = bad + 1
        bad_ids.append(student_id)

print(f"Valid: {valid}")
print(f"Bad: {bad}")
print(f"Bad IDs: {bad_ids}")
print(f"Average of valid scores: {total/valid:.1f}")
```

```python
#Exercise 5.2 — grades & top student among valid rows:
passed = 0
failed = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        name = parts[1]
        if score >= 50:
            passed = passed + 1
        else:
            failed = failed + 1
        if score > best_score:
            best_score = score
            best_name = name
    except ValueError:
        continue  # already counted as bad in 5.1

print(f"Passed: {passed}, Failed: {failed}")
print(f"Top student: {best_name} with {best_score}")
print(f"Their grade: {get_grade(best_score)}")
```

```python
#Exercise 5.3 — stretch: report by district (dictionary accumulators):
sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    district = parts[5]
    try:
        score = int(parts[7])
        sums[district] = sums.get(district, 0) + score
        counts[district] = counts.get(district, 0) + 1
    except ValueError:
        continue

for district in sums:
    avg = sums[district] / counts[district]
    print(f"{district}: average {avg:.1f} ({counts[district]} students)")
```

```python
# — write and read back a report file:
with open("week3_report.txt", "w") as f:
    f.write(f"Students processed: {valid + bad}\n")
    f.write(f"Valid: {valid}, Bad: {bad}\n")
    f.write(f"Average score: {total/valid:.1f}\n")
    f.write(f"Top student: {best_name} ({best_score})\n")

with open("week3_report.txt") as f:
    print(f.read())
```
