```python
def grade(score):
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

test_scores = [91, 76, 64, 52, 43]

for score in test_scores:
    print(score, "->", grade(score))
```

```python
def validation_result(score, attendance):
    if score >= 50 and attendance >= 75:
        return "Validated"
    elif score >= 50 and attendance < 75:
        return "Blocked: low attendance"
    else:
        return "Failed"

tests = [(76, 92), (76, 68), (43, 95)]

for score, attendance in tests:
    print((score, attendance), "->", validation_result(score, attendance))
```

```python
def get_grade(score):
    """Return the letter grade for a score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

for score in [91, 76, 43]:
    print(score, "->", get_grade(score))
```

```python
def pass_rate(score_list):
    passed = 0
    count = 0

    for score in score_list:
        count += 1
        if score >= 50:
            passed += 1

    return (passed / count) * 100

print("Pass rate:", pass_rate(scores))
```

```python
total = 0
valid_count = 0
bad_count = 0
bad_ids = []

for _, row in df.iterrows():
    try:
        s = int(row["score"])
        total += s
        valid_count += 1
    except (ValueError, TypeError):
        bad_count += 1
        bad_ids.append(row["student_id"])

average = total / valid_count

print("Valid:", valid_count)
print("Bad:", bad_count)
print("Bad IDs:", bad_ids)
print("Average of valid scores:", round(average, 1))
```

```python
passed = 0
failed = 0
best_score = -1
best_name = ""

for _, row in df.iterrows():
    try:
        s = int(row["score"])
    except (ValueError, TypeError):
        continue

    if s >= 50:
        passed += 1
    else:
        failed += 1

    if s > best_score:
        best_score = s
        best_name = row["name"]

print("Passed:", passed)
print("Failed:", failed)
print("Best score:", best_score)
print("Top student:", best_name)
print("Top student's grade:", get_grade(best_score))
```

```python
Exercise 5.3
```

```python
sums = {}
counts = {}

for _, row in df.iterrows():
    try:
        s = int(row["score"])
    except (ValueError, TypeError):
        continue

    d = row["district"]
    sums[d] = sums.get(d, 0) + s
    counts[d] = counts.get(d, 0) + 1

highest_district = ""
highest_average = -1

for d in sorted(sums):
    average = sums[d] / counts[d]
    print(d, "average:", round(average, 2))

    if average > highest_average:
        highest_average = average
        highest_district = d

print("District with highest average:", highest_district)
print("Highest average:", round(highest_average, 2))
```
