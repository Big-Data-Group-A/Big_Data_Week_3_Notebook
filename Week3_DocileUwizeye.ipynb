```python

df = pd.read_excel("week3_students.xlsx")


def get_grade(score):
    if score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"


total = 0
valid = 0
bad = 0
bad_ids = []

passed = 0
failed = 0

best_score = -1
best_name = ""


for index, row in df.iterrows():

    try:
        score = int(row.iloc[7])

        total += score
        valid += 1

        # Pass / fail
        if score >= 50:
            passed += 1
        else:
            failed += 1

        # Find highest score
        if score > best_score:
            best_score = score
            best_name = row.iloc[1]

    except ValueError:
        bad += 1
        bad_ids.append(row.iloc[0])


print("Passed:", passed)
print("Failed:", failed)
print("Top student:", best_name)
print("Top score:", best_score)
print("Top student's grade:", get_grade(best_score))
```

```python

df = pd.read_excel("week3_students.xlsx")

sums = {}
counts = {}

for index, row in df.iterrows():

    try:
        score = int(row.iloc[7])
        district = row.iloc[5]

        sums[district] = sums.get(district, 0) + score
        counts[district] = counts.get(district, 0) + 1

    except ValueError:
        continue


highest_district = ""
highest_average = -1

for district in sums:
    average = sums[district] / counts[district]

    print(district, "average:", round(average, 1))

    if average > highest_average:
        highest_average = average
        highest_district = district


print()
print("District with highest average:", highest_district)
print("Highest district average:", round(highest_average, 1))
```
