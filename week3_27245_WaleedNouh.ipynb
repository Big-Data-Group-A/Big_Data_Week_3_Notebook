```python
names = []

for line in lines[1:]:
    parts = line.strip().split(",")
    names.append(parts[1])

print("Count:", len(names))
print("First 5 names:", names[:5])
```

```python
valid = 0
bad = 0
bad_ids = []
total = 0

for row in lines[1:]:
    parts = row.strip().split(",")
    
    try:
        score = int(parts[7])
        total = total + score
        valid = valid + 1
    except ValueError:
        bad = bad + 1
        bad_ids.append(parts[0])

average = total / valid

print("Valid:", valid)
print("Bad:", bad)
print("Bad IDs:", bad_ids)
print("Average:", average)
```

```python
passed = 0
failed = 0
best_score = -1
best_name = ""

for row in lines[1:]:
    parts = row.strip().split(",")
    
    try:
        score = int(parts[7])
        
        if score >= 50:
            passed = passed + 1
        else:
            failed = failed + 1
        
        if score > best_score:
            best_score = score
            best_name = parts[1]
            
    except ValueError:
        pass

print("Passed:", passed)
print("Failed:", failed)
print("Top student:", best_name)
print("Top score:", best_score)
print("Top grade:", get_grade(best_score))
```

```python
sums = {}
counts = {}

for row in lines[1:]:
    parts = row.strip().split(",")
    
    try:
        score = int(parts[7])
        district = parts[5]
        
        sums[district] = sums.get(district, 0) + score
        counts[district] = counts.get(district, 0) + 1
        
    except ValueError:
        pass

for district in sums:
    average = sums[district] / counts[district]
    print(district, average)
```
