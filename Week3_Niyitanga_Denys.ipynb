## Part 1 — Making Decisions
### Exercise 1.1 — The grading ladder

```python
scores_to_test = [91, 76, 64, 52, 43]

for score in scores_to_test:
    if score >= 85:
        grade = "A"
    elif score >= 70:
        grade = "B"
    elif score >= 60:
        grade = "C"
    elif score >= 50:
        grade = "D"
    else:
        grade = "F"
    print(f"Score {score} -> Grade {grade}")
```

### Exercise 1.2 — Combined conditions
A course is validated only if `score >= 50` **and** `attendance >= 75`.

```python
tests = [(76, 92), (76, 68), (43, 95)]

for score, attendance in tests:
    if score >= 50 and attendance >= 75:
        print("Validated")
    elif score >= 50 and attendance < 75:
        print("Blocked: low attendance")
    else:
        print("Failed")
```

## Part 2 — Loops & the Accumulator
### Exercise 2.1 — Loop + filter

```python
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

for score in scores:
    if score >= 80:
        print(f"Top score: {score}")
```

### Exercise 2.2 — Accumulator pattern
No `sum()`, `len()`, or `max()` are used here.

```python
total = 0
count = 0
passed = 0
failed = 0

for score in scores:
    total = total + score
    count = count + 1
    if score >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

average = total / count
print(f"Total: {total}")
print(f"Count: {count}")
print(f"Average: {average:.1f}")
print(f"Passed: {passed}")
print(f"Failed: {failed}")
```

## Part 3 — Functions
### Exercise 3.1 — `get_grade`

```python
def get_grade(score):
    """Convert a score to a letter grade."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    return "F"

print(get_grade(91))
print(get_grade(76))
print(get_grade(43))
```

### Exercise 3.2 — `pass_rate`

```python
def pass_rate(score_list):
    """Return the percentage of scores that are 50 or above."""
    passed = 0
    count = 0

    for score in score_list:
        count = count + 1
        if score >= 50:
            passed = passed + 1

    return (passed / count) * 100

print(f"Pass rate: {pass_rate(scores):.1f}%")
```

## Part 5 — The Resilient Pipeline
### Exercise 5.1 — Survive the mess

```python
total = 0
valid = 0
bad = 0
bad_ids = []

with open("week3_students.csv") as f:
    lines = f.readlines()

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        total = total + score
        valid = valid + 1
    except ValueError:
        bad = bad + 1
        bad_ids.append(parts[0])

average_valid = total / valid
print(f"Valid: {valid}")
print(f"Bad: {bad}")
print(f"Bad IDs: {bad_ids}")
print(f"Average valid score: {average_valid:.1f}")
```

### Exercise 5.2 — Grades and the top student

```python
passed = 0
failed = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
    except ValueError:
        continue

    if score >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

    if score > best_score:
        best_score = score
        best_name = parts[1]

print(f"Passed: {passed}")
print(f"Failed: {failed}")
print(f"Top student: {best_name}")
print(f"Top score: {best_score}")
print(f"Top student grade: {get_grade(best_score)}")
```

### Exercise 5.3 — District averages

```python
sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    d = parts[5]
    try:
        s = int(parts[7])
    except ValueError:
        continue

    sums[d] = sums.get(d, 0) + s
    counts[d] = counts.get(d, 0) + 1

print("District averages:")
for d in sums:
    print(f"{d}: {sums[d] / counts[d]:.1f}")
```
