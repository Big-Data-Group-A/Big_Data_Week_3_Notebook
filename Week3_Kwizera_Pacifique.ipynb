# Exercise 1.1 - The grading ladder

```python
# Exercise 1.1 - The grading ladder
test_scores = [91, 76, 64, 52, 43]

for score in test_scores:
  if score >= 85:
    grade = 'A'
  elif score >= 70:
    grade = 'B'
  elif score >= 60:
    grade = 'C'
  elif score >= 50:
    grade = 'D'
  else:
    grade = 'F'
  print(f'Score: {score} -> Grade: {grade}')

print('-' * 40)

# Exercise 1.2 - Combined conditions
test_cases = [(76, 92), (76, 68), (43, 95)]

for score, attendance in test_cases:
  if score >= 50 and attendance >= 75:
    status = 'Validated'
  elif score >= 50 and attendance < 75:
    status = 'Blocked: low attendance'
  else:
    status = 'Failed'
  print(
      f'Score: {score}, Attendance: {attendance}% -> Decision: {status}'
  )
```

```python
# Exercise 3.1 - get_grade
def get_grade(score):
  """Returns the letter grade (A, B, C, D, F) corresponding to a numerical score."""
  if score >= 85:
    return "A"
  elif score >= 70:
    return "B"
  elif score >= 60:
    return "C"
  elif score >= 50:
    return "D"
  else:
    return "F"


# Test 3.1
for s in [91, 76, 43]:
  print(f"Score {s} -> {get_grade(s)}")

print("-" * 40)


# Exercise 3.2 - pass_rate
def pass_rate(score_list):
  """Returns the percentage of scores greater than or equal to 50."""
  passed_count = 0
  total_count = 0
  for score in score_list:
    total_count += 1
    if score >= 50:
      passed_count += 1
  return (passed_count / total_count) * 100


# Test 3.2
print(f"Pass Rate: {pass_rate(scores)}%")  # Expected output: 90.0
```

# Exercise 5.2 - Grades & the top student
# Exercise 5.3 - Stretch: report by district

```python
# Exercise 5.1 - Survive the mess
valid_count = 0
bad_count = 0
total_valid_score = 0
bad_ids = []

for line in lines[1:]:
  parts = line.strip().split(',')
  student_id = parts[0]
  score_str = parts[7]

  try:
    score = int(score_str)
    total_valid_score += score
    valid_count += 1
  except ValueError:
    bad_count += 1
    bad_ids.append(student_id)

valid_avg = round(total_valid_score / valid_count, 1)

print(f'Valid records count: {valid_count}')
print(f'Bad records count: {bad_count}')
print(f'Bad IDs: {bad_ids}')
print(f'Average valid score: {valid_avg}')

print('-' * 40)

# Exercise 5.2 - Grades & the top student
passed_valid = 0
failed_valid = 0
best_score = -1
best_name = ''

for line in lines[1:]:
  parts = line.strip().split(',')
  name = parts[1]
  score_str = parts[7]

  try:
    score = int(score_str)
    if score >= 50:
      passed_valid += 1
    else:
      failed_valid += 1

    if score > best_score:
      best_score = score
      best_name = name
  except ValueError:
    continue

print(f'Passed valid students: {passed_valid}')
print(f'Failed valid students: {failed_valid}')
print(
    f'Top Student: {best_name} with Score: {best_score} (Grade:'
    f' {get_grade(best_score)})'
)

print('-' * 40)

# Exercise 5.3 - Stretch: report by district
sums = {}
counts = {}

for line in lines[1:]:
  parts = line.strip().split(',')
  district = parts[5]
  score_str = parts[7]

  try:
    score = int(score_str)
    sums[district] = sums.get(district, 0) + score
    counts[district] = counts.get(district, 0) + 1
  except ValueError:
    continue

print('District Averages:')
highest_avg = -1
top_district = ''

for district in sums:
  avg = sums[district] / counts[district]
  print(f'  {district}: {avg:.2f}')
  if avg > highest_avg:
    highest_avg = avg
    top_district = district

print(
    f'\nDistrict with the highest average score: {top_district} ({highest_avg:.2f})'
)
```

```python
# Writing to week3_report.txt
report_content = f"""Week 3 Student Data Analysis Report
===================================
Total Student Records Processed: {len(lines) - 1}
Valid Records: {valid_count}
Bad Records: {bad_count} (IDs: {', '.join(bad_ids)})
Average Score (Valid): {valid_avg}
Top Student: {best_name} (Score: {best_score}, Grade: {get_grade(best_score)})
Highest Average District: {top_district} ({highest_avg:.2f})
"""

with open("week3_report.txt", "w") as f:
  f.write(report_content)

# Reading back the file to verify
with open("week3_report.txt", "r") as f:
  print(f.read())
```

### 1. Data Trust Threshold & Strategy
**Question:** Your pipeline skipped 3 broken records out of 40 (7.5%). At what percentage would you stop trusting the dataset — and what would you do instead of skipping?

**Answer:**
At a 7.5% corruption rate, skipping unusable rows is acceptable for initial exploratory analysis without heavily biasing overall trends. However, if corrupt or missing records reach **15% or higher**, I would stop trusting the dataset for direct automated consumption because dropping that many records introduces significant selection bias.
Instead of simply skipping:
1. **Source Audit & Upstream Validation:** Pause the pipeline and inspect data collection logs, database extraction queries, and ingestion pipelines to identify why records are missing or corrupted.
2. **Imputation Techniques:** Where appropriate, apply statistical data imputation (such as replacing missing scores with the district average or class median, or using predictive models).
3. **Dead-Letter Queue:** Route all malformed records to an error log/quarantine file for manual human review and correction.
