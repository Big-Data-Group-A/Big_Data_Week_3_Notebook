## prt 6 · Reflection**1. Is this 20-student dataset Big Data? Which of the 5 V's does it clearly not have yet?**This dataset is not Big Data. It only contains 20 records, so it clearly lacks **Volume** — one of the core 5 V's(Volume, Velocity, Variety, Veracity, Value). Big Data usually refers to datasets so large that they can't be opened,viewed, or processed with everyday tools like Excel, whereas this entire table can be scrolled through and read inseconds. It's also missing **Velocity**, since it's a single static file rather than data continuously streaming inreal time (like sensor readings or social media posts). To be considered Big Data, this dataset would need to scaleup to thousands or millions of student records collected continuously over time.**2.
