```python
# Exercise 2.2
total = 0
count = 0
passed = 0
failed = 0

for s in scores:
    total = total + s
    count = count + 1
    if s >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

average = total / count

print("Total:", total)
print("Count:", count)
print("Average:", average)
print("Passed:", passed)
print("Failed:", failed)
```

```python
# Part 3
# Exercise 3.1
def get_grade(score):
    """Return the letter grade (A-F) for a given numeric score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

# Test it
print(get_grade(91))
print(get_grade(76))
print(get_grade(43))
```

```python
# Exercise 3.2
def pass_rate(score_list):
    """Return the percentage of scores in score_list that are >= 50."""
    passed = 0
    total = 0
    
    for s in score_list:
        total = total + 1
        if s >= 50:
            passed = passed + 1
    
    percentage = (passed / total) * 100
    return percentage

# Test it on the Part 2 list
print(int(pass_rate(scores)))
```

```python
# Part 4
# Exercise 4.1
import pandas as pd
df = pd.read_excel("week3_students.xlsx")
df.to_csv("week3_students.csv", index=False)
```

```python
with open("week3_students.csv") as f:
    lines = f.readlines()

print("Number of lines:", len(lines))
print("Header line:", lines[0])
print("First student's line:", lines[1])
# Why 41 lines, not 40?
# Because line 0 is the header row (column names), and the 40 students start from line 1 onward.
```

```python
# Exercise 4.2
names = []

for line in lines[1:]:
    parts = line.strip().split(",")
    names.append(parts[1])

print("Count:", len(names))
print("First 5 names:", names[:5])
```

```python
# Part 5
# Exercise 5.1
total = 0
valid_count = 0
bad_count = 0
bad_ids = []

for line in lines[1:]:
    parts = line.strip().split(",")
    student_id = parts[0]
    score_str = parts[7]
    
    try:
        score = int(score_str)
        total = total + score
        valid_count = valid_count + 1
    except ValueError:
        bad_count = bad_count + 1
        bad_ids.append(student_id)

average = round(total / valid_count, 1)

print("Valid:", valid_count)
print("Bad:", bad_count)
print("Bad IDs:", bad_ids)
print("Average of valid scores:", average)
```

```python
# Exercise 5.2
passed = 0
failed = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    name = parts[1]
    score_str = parts[7]
    
    try:
        score = int(score_str)
    except ValueError:
        continue  # skip broken records, already counted in 5.1
    
    # (a) passed/failed count
    if score >= 50:
        passed = passed + 1
    else:
        failed = failed + 1
    
    # (b) track the best score
    if score > best_score:
        best_score = score
        best_name = name

print("Passed:", passed)
print("Failed:", failed)
print("Top student:", best_name, "with score", best_score)
print("Top student's grade:", get_grade(best_score))
```

```python
# Exercise 5.3
sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    district = parts[5]
    score_str = parts[7]
    
    try:
        score = int(score_str)
    except ValueError:
        continue  # skip broken records
    
    sums[district] = sums.get(district, 0) + score
    counts[district] = counts.get(district, 0) + 1

# Print every district's average
print("District averages:")
for district in sums:
    avg = sums[district] / counts[district]
    print(district, ":", round(avg, 1))

# Find the district with the highest average
best_district = ""
best_avg = -1

for district in sums:
    avg = sums[district] / counts[district]
    if avg > best_avg:
        best_avg = avg
        best_district = district

print()
print("Highest average district:", best_district, "with average", round(best_avg, 1))
```

```python
# Exercise 5.4
with open("week3_report.txt", "w") as f:
    f.write("Student count: 40\n")
    f.write("Valid: " + str(valid_count) + "\n")
    f.write("Bad: " + str(bad_count) + "\n")
    f.write("Average score: " + str(average) + "\n")
    f.write("Top student: " + best_name + " with score " + str(best_score) + "\n")

# Read it back
with open("week3_report.txt", "r") as f:
    print(f.read())
```
