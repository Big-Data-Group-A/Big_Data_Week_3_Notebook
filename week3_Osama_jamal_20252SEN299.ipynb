```python
with open("week3_students.csv") as f:
    lines = f.readlines()

print("Number of lines:", len(lines))
print("Header:", lines[0].strip())
print("First student:", lines[1].strip())
```

```python
names = []
for line in lines[1:]:
    parts = line.strip().split(",")
    names.append(parts[1])

print("Count:", len(names))
print("First 5:", names[:5])
```

```python
valid = 0
bad = 0
bad_ids = []
total = 0

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])          # risky line: may raise ValueError
        total = total + score          # only reached if int() worked
        valid = valid + 1
    except ValueError:
        bad = bad + 1
        bad_ids.append(parts[0])       # remember which student_id was broken

average = total / valid

print("Valid:", valid)
print("Bad:", bad)
print("Bad IDs:", bad_ids)
print("Average of valid scores:", round(average, 1))
```

```python
passed = 0
failed = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        if score >= 50:
            passed = passed + 1
        else:
            failed = failed + 1
        if score > best_score:
            best_score = score
            best_name = parts[1]
    except ValueError:
        pass                           # broken row: ignore it (valid rows only)

print("Passed:", passed)
print("Failed:", failed)
print("Top student:", best_name, "with", best_score)
print("Top student's grade:", get_grade(best_score))
```

```python
sums = {}       # district -> running total of scores
counts = {}     # district -> how many valid scores

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        s = int(parts[7])
        d = parts[5]
        sums[d] = sums.get(d, 0) + s
        counts[d] = counts.get(d, 0) + 1
    except ValueError:
        pass                                   # broken row: skip it

best_district = ""
best_avg = -1

for d in sums:
    avg = sums[d] / counts[d]
    print(d, "->", round(avg, 1), "| students:", counts[d])
    if avg > best_avg:
        best_avg = avg
        best_district = d

print("Highest average:", best_district, round(best_avg, 1))
```
