```python
# Exercise 1.1(grading) 

scores = [91, 76, 64, 52, 43]

for score in scores:
    if score >= 85:
        grade = "A"
    elif score >= 70:
        grade = "B"
    elif score >= 60:
        grade = "C"
    elif score >= 50:
        grade = "D"
    else:
        grade = "F"

    print(score, "->", grade)
```

```python
# Exercise 1.2  (Combining)

tests = [
    (76, 92),
    (76, 68),
    (43, 95)
]

for score, attendance in tests:
    if score >= 50 and attendance >= 75:
        print("Validated")
    elif score >= 50 and attendance < 75:
        print("Blocked: low attendance")
    else:
        print("Failed")
```

```python
# Exercise 2.1 (top score)

scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

for score in scores:
    if score >= 80:
        print("Top score:", score)
```

```python
# Exercise 2.2 (to Accumulate)

scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

total = 0
count = 0
passed = 0
failed = 0

for score in scores:
    total = total + score
    count = count + 1

    if score >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

average = total / count

print("Total:", total)
print("Count:", count)
print("Average:", average)
print("Passed:", passed)
print("Failed:", failed)
```

```python
# Exercise 2.3 (using while loop)

saved = 0
monthly_saving = 15000
target = 250000
months = 0

while saved < target:
    saved = saved + monthly_saving
    months = months + 1

print("Months needed:", months)
print("Total saved:", saved)
```

```python
# Exercise 3.1 (grading)

def get_grade(score):
    """Return the letter grade for a score."""
    
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"


print(get_grade(91))
print(get_grade(76))
print(get_grade(43))
```

```python
# Exercise 3.2 (passing rate )

def pass_rate(score_list):
    """Return the percentage of scores that are 50 or higher."""
    
    passed = 0
    count = 0

    for score in score_list:
        count = count + 1

        if score >= 50:
            passed = passed + 1

    return (passed / count) * 100


scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

print("Pass rate:", pass_rate(scores))
```

```python
with open("week3_students.csv", "r") as f:
    lines = f.readlines()

print("Number of lines:", len(lines))
print("Header:", lines[0].strip())
print("First student:", lines[1].strip())
```

```python
# Exercise 4.2 (to Extract names)

names = []

for line in lines[1:]:
    parts = line.strip().split(",")
    names.append(parts[1])

print("Number of names:", len(names))
print("First 5 names:", names[:5])
```

```python
# Exercise 5.1  (bad data)

valid = 0
bad = 0
bad_ids = []
total = 0

for line in lines[1:]:
    parts = line.strip().split(",")

    try:
        score = int(parts[7])
        total += score
        valid += 1
    except ValueError:
        bad += 1
        bad_ids.append(parts[0])

print("Valid records:", valid)
print("Bad records:", bad)
print("Bad student IDs:", bad_ids)
print("Average valid score:", round(total / valid, 1))
```

```python
# Exercise 5.2 (Pass/fail student)

passed = 0
failed = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")

    try:
        score = int(parts[7])
        
        if score >= 50:
            passed += 1
        else:
            failed += 1

        if score > best_score:
            best_score = score
            best_name = parts[1]

    except ValueError:
        continue

print("Passed:", passed)
print("Failed:", failed)
print("Top student:", best_name)
print("Top score:", best_score)
print("Grade:", get_grade(best_score))
```

```python
# Exercise 5.3 — District averages

sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")

    try:
        score = int(parts[7])
        district = parts[5]

        sums[district] = sums.get(district, 0) + score
        counts[district] = counts.get(district, 0) + 1

    except ValueError:
        continue

for district in sums:
    average = sums[district] / counts[district]
    print(district, ":", round(average, 1))
```
