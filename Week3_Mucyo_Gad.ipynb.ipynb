```python
# Exercise 1.1 — The grading ladder
print("=== Exercise 1.1 ===")

def print_grade(score):
    if score >= 85:
        grade = "A"
    elif score >= 70:
        grade = "B"
    elif score >= 60:
        grade = "C"
    elif score >= 50:
        grade = "D"
    else:
        grade = "F"
    print(f"Score {score} -> Grade {grade}")

for test_score in [91, 76, 64, 52, 43]:
    print_grade(test_score)

# Exercise 1.2 — Combined conditions
print("\n=== Exercise 1.2 ===")

def check_validation(score, attendance):
    if score >= 50 and attendance >= 75:
        result = "Validated"
    elif score >= 50 and attendance < 75:
        result = "Blocked: low attendance"
    else:
        result = "Failed"
    print(f"score={score}, attendance={attendance} -> {result}")

check_validation(76, 92)
check_validation(76, 68)
check_validation(43, 95)

# Question 1.3 (answer as a comment):
# If "elif score >= 50" came before "elif score >= 70", every score of 70+
# would already match the >=50 branch and get grade D, so A, B and C
# would never be reached.
```

```python
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]  # given for Part 2

# Exercise 2.1 — Loop + filter
print("=== Exercise 2.1 ===")
for s in scores:
    if s >= 80:
        print(f"Top score: {s}")

# Exercise 2.2 — The accumulator pattern (no sum/len/max)
print("\n=== Exercise 2.2 ===")
total = 0
count = 0
passed = 0
failed = 0

for s in scores:
    total = total + s
    count = count + 1
    if s >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

average = total / count
print(f"Total: {total}")
print(f"Count: {count}")
print(f"Average: {average}")
print(f"Passed: {passed}")
print(f"Failed: {failed}")

# Exercise 2.3 — while loop
print("\n=== Exercise 2.3 ===")
savings_per_month = 15000
target = 250000
saved = 0
months = 0

while saved < target:
    saved = saved + savings_per_month
    months = months + 1

print(f"Months needed: {months}")
```

```python
# Exercise 3.1 — get_grade
print("=== Exercise 3.1 ===")

def get_grade(score):
    """Return the letter grade (A/B/C/D/F) for a given numeric score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

for test_score in [91, 76, 43]:
    print(f"{test_score} -> {get_grade(test_score)}")

# Exercise 3.2 — pass_rate
print("\n=== Exercise 3.2 ===")

def pass_rate(score_list):
    """Return the percentage of scores in score_list that are >= 50."""
    total_count = 0
    passed_count = 0
    for s in score_list:
        total_count = total_count + 1
        if s >= 50:
            passed_count = passed_count + 1
    return (passed_count / total_count) * 100

print(f"Pass rate: {pass_rate(scores)}")
```

```python
# Exercise 4.1 — First contact
print("=== Exercise 4.1 ===")

with open("week3_students.csv") as f:
    lines = f.readlines()

print(f"Number of lines: {len(lines)}")
print(f"Header line: {lines[0].strip()}")
print(f"First student's line: {lines[1].strip()}")
# There are 41 lines, not 40, because line 1 is the header (column names),
# not a student record.

# Exercise 4.2 — Extract a column
print("\n=== Exercise 4.2 ===")

names = []
for line in lines[1:]:
    parts = line.strip().split(",")
    names.append(parts[1])

print(f"Number of names: {len(names)}")
print(f"First 5 names: {names[:5]}")
```

```python
# Exercise 5.1 — Survive the mess
print("=== Exercise 5.1 ===")

valid = 0
bad = 0
bad_ids = []
total_valid_scores = 0

for line in lines[1:]:
    parts = line.strip().split(",")
    student_id = parts[0]
    try:
        score = int(parts[7])
        valid = valid + 1
        total_valid_scores = total_valid_scores + score
    except ValueError:
        bad = bad + 1
        bad_ids.append(student_id)

average_valid = total_valid_scores / valid

print(f"Valid: {valid}")
print(f"Bad: {bad}")
print(f"Bad IDs: {bad_ids}")
print(f"Average of valid scores: {round(average_valid, 1)}")

# Exercise 5.2 — Grades & the top student
print("\n=== Exercise 5.2 ===")

passed_valid = 0
failed_valid = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
    except ValueError:
        continue

    name = parts[1]
    if score >= 50:
        passed_valid = passed_valid + 1
    else:
        failed_valid = failed_valid + 1

    if score > best_score:
        best_score = score
        best_name = name

print(f"Passed: {passed_valid}")
print(f"Failed: {failed_valid}")
print(f"Top student: {best_name} with score {best_score}")
print(f"Top student's grade: {get_grade(best_score)}")

# Exercise 5.3 — Stretch: report by district
print("\n=== Exercise 5.3 ===")

sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
    except ValueError:
        continue

    d = parts[5]
    sums[d] = sums.get(d, 0) + score
    counts[d] = counts.get(d, 0) + 1

best_district = ""
best_district_avg = -1

for d in sums:
    d_avg = sums[d] / counts[d]
    print(f"{d}: average = {round(d_avg, 1)}")
    if d_avg > best_district_avg:
        best_district_avg = d_avg
        best_district = d

print(f"\nDistrict with the highest average score: {best_district} ({round(best_district_avg, 1)})")
```
