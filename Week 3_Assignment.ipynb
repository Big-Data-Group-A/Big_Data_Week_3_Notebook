```python
def_scores = [91, 76, 64, 52, 43]     

for score in def_scores:              
    if score >= 85:                   
        grade = "A"
    elif score >= 70:                 
        grade = "B"
    elif score >= 60:                
        grade = "C"
    elif score >= 50:                 
        grade = "D"
    else:                             
        grade = "F"
    print(f"Score {score} -> Grade {grade}")
```

## Part 3

## 3.1

```python
def get_grade(score):
    """Convert a numeric score to a letter grade A-F."""
    if score >= 85:
        return "A"                   
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    return "F"                        

print(get_grade(91), get_grade(76), get_grade(43)) 
```

## 3.2

```python
def pass_rate(score_list):
    """Return the percentage of scores that are >= 50."""
    passed = 0                        
    for s in score_list:
        if s >= 50:
            passed = passed + 1
    return passed / len(score_list) * 100

print(pass_rate(scores))  
```

## 4.2

```python
names = []
for line in lines[1:]:                  
    parts = line.strip().split(",")     
    names.append(parts[1])              
print(len(names))                       
print(names[:5])    
```

## Part 5

## 5.1

```python
get_grade_ready = get_grade              

total = 0; valid = 0; bad = 0
bad_ids = []
passed = 0; failed = 0
best_score = -1; best_name = ""
sums = {}; counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        s = int(parts[7])                
    except ValueError:
        bad = bad + 1                    
        bad_ids.append(parts[0])         
        continue                         
    
    total = total + s
    valid = valid + 1

    if s >= 50: passed = passed + 1      
    else:       failed = failed + 1

    if s > best_score:                  
        best_score = s
        best_name = parts[1]

    d = parts[5]                        
    sums[d] = sums.get(d, 0) + s         
    counts[d] = counts.get(d, 0) + 1


print(f"Valid: {valid}, Bad: {bad}, Bad IDs: {bad_ids}")
print(f"Average: {total/valid:.1f}")
```

## 5.2 

```python
print(f"Passed: {passed}, Failed: {failed}")
print(f"Top: {best_name} with {best_score} -> grade {get_grade(best_score)}")
```

```python
for d in sums:
    print(f"{d}: {sums[d]/counts[d]:.1f}")
best_d = max(sums, key=lambda d: sums[d] / counts[d])
print("Highest district:", best_d)
```
