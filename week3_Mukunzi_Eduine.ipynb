Step 1: Part 1 — Making Decisions (5 marks)
Exercise 1.1 — The grading ladder
Write an if/elif/else ladder to classify a score into letter grades:
$A \ge 85$, $B: 70–84$, $C: 60–69$, $D: 50–59$, $F < 50$

```python
# Exercise 1.1
scores_to_test = [91, 76, 64, 52, 43]

for score in scores_to_test:
    if score >= 85:
        grade = 'A'
    elif score >= 70:
        grade = 'B'
    elif score >= 60:
        grade = 'C'
    elif score >= 50:
        grade = 'D'
    else:
        grade = 'F'
    print(f"Score {score}: Grade {grade}")
```

Exercise 1.2 — Combined conditions
Evaluate course validation where a student validates only if score >= 50 and attendance >= 75

```python
# Exercise 1.2
test_cases = [(76, 92), (76, 68), (43, 95)]

for score, att in test_cases:
    if score >= 50 and att >= 75:
        status = "Validated"
    elif score >= 50 and att < 75:
        status = "Blocked: low attendance"
    else:
        status = "Failed"
    print(f"Score: {score}, Attendance: {att}% -> {status}")
```

Question 1.3 — Answer: If elif score >= 50: came before elif score >= 70:, any score of $70$ or higher (like $85$) would evaluate as true for >= 50 first, incorrectly triggering that branch and preventing the >= 70 condition from ever running.

```python
# Exercise 2.2
total = 0
count = 0
passed_count = 0
failed_count = 0

for s in scores:
    total += s
    count += 1
    if s >= 50:
        passed_count += 1
    else:
        failed_count += 1

avg = total / count if count > 0 else 0

print(f"Total: {total}")
print(f"Count: {count}")
print(f"Average: {avg}")
print(f"Passed count: {passed_count}")
print(f"Failed count: {failed_count}")
```

Exercise 3.2 — pass_rate

```python
# Exercise 3.2
def pass_rate(score_list):
    """Returns the percentage of scores greater than or equal to 50."""
    passed = 0
    total = 0
    for s in score_list:
        total += 1
        if s >= 50:
            passed += 1
    return (passed / total) * 100 if total > 0 else 0.0

print(f"Pass rate: {pass_rate(scores)}%") # Expected output: 90.0
```

Step 5: Part 5 — The Resilient Pipeline 
Exercise 5.1 — Survive the mess

```python
# Exercise 5.1
total_score = 0
valid_count = 0
bad_count = 0
bad_ids = []

for line in lines[1:]:
    parts = line.strip().split(",")
    student_id = parts[0]
    score_str = parts[7]
    
    try:
        score = int(score_str)
        total_score += score
        valid_count += 1
    except ValueError:
        bad_count += 1
        bad_ids.append(student_id)

avg_valid_score = total_score / valid_count if valid_count > 0 else 0

print(f"Valid count: {valid_count}")
print(f"Bad count: {bad_count}")
print(f"Bad student IDs: {bad_ids}")
print(f"Average of valid scores: {avg_valid_score:.1f}")
```

Exercise 5.2 — Grades & the top student

```python
# Exercise 5.2
passed_count = 0
failed_count = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    name = parts[1]
    
    try:
        score = int(parts[7])
        if score >= 50:
            passed_count += 1
        else:
            failed_count += 1
            
        if score > best_score:
            best_score = score
            best_name = name
    except ValueError:
        continue

print(f"Passed count: {passed_count}")
print(f"Failed count: {failed_count}")
print(f"Top student: {best_name} with score {best_score}")
print(f"Top student grade: {get_grade(best_score)}")
```

Exercise 5.3 — Report by district

```python
# Exercise 5.3
sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    district = parts[5]
    
    try:
        score = int(parts[7])
        sums[district] = sums.get(district, 0) + score
        counts[district] = counts.get(district, 0) + 1
    except ValueError:
        continue

district_averages = {}
print("District Averages:")
for d in sums:
    avg_d = sums[d] / counts[d]
    district_averages[d] = avg_d
    print(f"  {d}: {avg_d:.2f}")

highest_district = max(district_averages, key=district_averages.get)
print(f"\nDistrict with the highest average score: {highest_district} ({district_averages[highest_district]:.2f})")
```
