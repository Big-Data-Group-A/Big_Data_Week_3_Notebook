```python
#Part 3
#Exercise 3.1

def get_grade(score):
    """Return the letter grade for a score."""
    if score >=85:
        return'A'
    elif score >=70:
        return 'B'
    elif score >=60:
        return  'C'
    elif score >= 50:
        return 'D'
    else:
        return 'F'

print(get_grade.__doc__)
print(f"Grade: {get_grade(91)}")
print(f"Grade: {get_grade(76)}")
print(f"Grade: {get_grade(43)}")
```

```python
#Part 3
#Exercise 3.2

def pass_rate(score_list):
    passed=0
    totalStudents =0
    for score in score_list:
        totalStudents +=1
        if score >=50:
            passed +=1
    percentage = (passed/totalStudents) * 100

    return percentage


print(f"Percentage: {pass_rate(scores)}")
```

```python
#Part 4
#Exercise 4.1

with open("week3_students.csv") as f:
    lines = f.readlines()
    print("Total lines: ",len(lines))
    print("Header line: ",lines[0])
    print("First student line: ",lines[1])

    # It is 41 lines because we counted all libes including a header
```

```python
#Part 4
#Exercise 4.2

names =[]
for line in lines[1:]:
    row=line.strip().split(",")
    names.append(row[1])



print("Total names: ", len(names))
print("First 5 names: ")
for name in range(5):
    print(names[name])

```

```python
# Part 5
# Exercise 5.1
total = 0
bad = 0
valid = 0
count = 0
bad_ids = []
for line in lines[1:]:
    count = count + 1
    parts = line.strip().split(",")
    try:
        total = total + int(parts[7])
        valid = valid + 1
    except ValueError:
        bad = bad + 1
        bad_ids.append(parts[0]) 
print("Total Valid: ", valid)
print("Total Bad: ", bad)
print("Bad ids: ", bad_ids)
print("Average: ", total/count)
```

```python
# Part 5
# Exercise 5.2

failed = 0
passed = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")

    try:
        score = int(parts[7])
        if score >=50:
            passed +=1
        if score <50:
            failed +=1
        if score > best_score:
            best_score = score
            best_name = parts[1]

    except ValueError:
        pass
print("Students Passed: ", passed)
print("Failed students: ", failed)
print("Highest Score: ", best_score)
print("Student with highest score: ", best_name)
print("Grade: ",get_grade(best_score))
```

```python
#Part 5
#Exercise 5.3
sums = {}
counts = {}
for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        district = parts[5]
        # Add the score to that district's total
        sums[district] = sums.get(district, 0) + score
        # Count students in that district
        counts[district] = counts.get(district, 0) + 1
    except ValueError:
        pass
# Print the average for each district
for district in sums:
    average = sums[district] / counts[district]
    print(district, ":", round(average, 1))
```

```python
highest_district = ""
highest_average = 0

for district in sums:
    average = sums[district] / counts[district]

    if average > highest_average:
        highest_average = average
        highest_district = district

print("District with highest average:", highest_district)
print("Highest average:", round(highest_average, 1))
```
