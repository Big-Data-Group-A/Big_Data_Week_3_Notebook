```python
#Exercise 1.1
scores = [91,76,64,52,43]
for score in scores:
    if score >= 85:
        grade = 'A'
    elif score >= 70:
        grade = 'B'
    elif score >= 60:
        grade = 'C'
    elif score >= 50:
        grade = 'D'
    else:
        grade = 'F'
    print(score, " ->" ,grade)
```

```python
#Exercise 2.2
total = 0
count = 0
passed = 0
failed = 0

for s in scores:
    total += s
    count += 1

    if s >= 50:
        passed += 1
    else:
        failed += 1

Avg = total / count
print('Total: ', total)
print('Count: ', count)
print('Average: ', Avg)
print('Passed: ', passed)
print('Failed: ', failed)
```

```python
#Exercise 3.1
def get_grade(score):
    if score >= 80:
        return 'A'
    elif score >= 70:
        return 'B'
    elif score >= 60:
        return 'C'
    elif score >= 50:
        return 'D'
    else:
        return 'F'
for s in [91,76,43]:
    print(s, get_grade(s))
```

```python
#Exercise 3.2
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]
def pass_rate(score_list):
    total = 0
    passed = 0

    for s in score_list:
        total += 1
        if s >= 50:
            passed += 1
    return passed / total * 100

print(pass_rate(scores))
```

```python
#Exercise 4.1
with open ("week3_students.csv") as f:
    lines = f.readlines()
    print('Number of lines: ', len(lines))
    print('Header: ', lines[0].strip())
    print('First student: ', lines[1].strip())

    # Why 41 lines, not 40?
    # Because the first line is the header (column names), not a student(1 header + 40 students = 41 lines).
```

```python
#Exercise 4.2
names = []
for line in lines[1:]:
    parts = line.strip().split(",")
    names.append(parts[1])

print('Count: ',len(names))
print('Names: ',names[:5])
```

```python
#Exercise 5.1
total = 0
valid = 0
bad = 0
bad_ids = []

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        s = int(parts[7])
    except ValueError:
        bad += 1
        bad_ids.append(parts[0])
        continue
    total += s
    valid += 1

print('Valid: ', valid)
print('Bad: ', bad)
print('Bad IDs: ', bad_ids)
print('Average of valid scores: ', round(total / valid, 1))
        
```

```python
#Exercise 5.2
passed = 0
failed = 0
best_score = -1
best_name = ''

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        s = int(parts[7])
    except ValueError:
        continue
    if s >= 50:
        passed += 1
    else:
        failed += 1
    if s > best_score:
        best_score = s
        best_name = parts[1]
print('Passed: ', passed, 'Failed: ', failed)
print('Top student: ', best_name, 'with ', best_score)
print('Their grade: ', get_grade(best_score))
```

```python
#Exercise 5.3
sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        s = int(parts[7])
    except ValueError:
        continue
    d = parts[5]
    sums[d] = sums.get(d,0) + s
    counts[d] = counts.get(d,0) + 1
for d in sums:
    print(d, round(sums[d] / counts[d], 1))

best_district = max(sums, key=lambda d: sums[d] / counts[d])
print('Highest Average: ', best_district)
```

```python
#Bonus Question
with open("week3_report.txt", "w") as f:
    f.write(f"Students: {len(lines) - 1}\n")
    f.write(f"Valid: {valid}, Bad: {bad}\n")
    f.write(f"Average: {total / valid: .1f}\n")
    f.write(f"Top student: {best_name} with {best_score}\n")

with open("week3_report.txt") as f:
    print(f.read())
```
