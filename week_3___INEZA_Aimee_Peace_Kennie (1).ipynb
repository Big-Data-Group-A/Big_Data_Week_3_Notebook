```python
#3.1
def get_grade(score):
    """Return the letter grade for a numeric score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

for s in [91, 76, 43]:
    print(s, "->", get_grade(s))
```

```python
#3.2
def pass_rate(score_list):
    passed = 0
    count = 0
    for s in score_list:
        count += 1
        if s >= 50:
            passed += 1
    return (passed / count) * 100

print(pass_rate(scores))
```

```python
#5.1
valid = 0
bad = 0
bad_ids = []
total_score = 0

with open("week3_students.csv", "r") as f:
    lines = f.readlines()

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        total_score += score
        valid += 1
    except ValueError:
        bad += 1
        bad_ids.append(parts[0])

avg_valid = total_score / valid

print("Valid:", valid)
print("Bad:", bad)
print("Bad IDs:", bad_ids)
print("Average valid score:", round(avg_valid, 1))
```

```python
#5.2
passed = 0
failed = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
    except ValueError:
        continue

    if score >= 50:
        passed += 1
    else:
        failed += 1

    if score > best_score:
        best_score = score
        best_name = parts[1]

print("Passed:", passed)
print("Failed:", failed)
print("Top student:", best_name, "-", best_score, "-", get_grade(best_score))
```

```python
#5.3
sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        s = int(parts[7])
    except ValueError:
        continue

    d = parts[5]
    sums[d] = sums.get(d, 0) + s
    counts[d] = counts.get(d, 0) + 1

for d in sums:
    print(d, round(sums[d] / counts[d], 1))

best_district = None
best_avg = -1

for d in sums:
    avg = sums[d] / counts[d]
    if avg > best_avg:
        best_avg = avg
        best_district = d

print("Highest district average:", best_district, round(best_avg, 1))
```
