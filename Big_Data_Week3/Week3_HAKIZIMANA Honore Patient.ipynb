```python
def get_grade(score):
    """Return the letter grade for a score."""
    
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"
```

```python
def pass_rate(score_list):
    """Return the percentage of scores that are 50 or higher."""
    
    passed = 0
    count = 0

    for score in score_list:
        count = count + 1

        if score >= 50:
            passed = passed + 1

    return (passed / count) * 100
```

```python
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

print(pass_rate(scores))
```

```python
total = 0
valid = 0
bad = 0
bad_ids = []

with open("week3_students.csv", "r") as f:
    lines = f.readlines()

for line in lines[1:]:
    parts = line.strip().split(",")

    student_id = parts[0]

    try:
        score = int(parts[7])

        total = total + score
        valid = valid + 1

    except ValueError:
        bad = bad + 1
        bad_ids.append(student_id)

average = total / valid

print("Valid:", valid)
print("Bad:", bad)
print("Bad IDs:", bad_ids)
print("Average:", round(average, 1))
```

```python
passed = 0
failed = 0

best_score = -1
best_name = ""

with open("week3_students.csv", "r") as f:
    lines = f.readlines()

for line in lines[1:]:
    parts = line.strip().split(",")

    name = parts[1]

    try:
        score = int(parts[7])

        if score >= 50:
            passed = passed + 1
        else:
            failed = failed + 1

        if score > best_score:
            best_score = score
            best_name = name

    except ValueError:
        continue

print("Passed:", passed)
print("Failed:", failed)
print("Best score:", best_score)
print("Top student:", best_name)
print("Top student's grade:", get_grade(best_score))
```

```python
sums = {}
counts = {}

with open("week3_students.csv", "r") as f:
    lines = f.readlines()

for line in lines[1:]:
    parts = line.strip().split(",")

    district = parts[5]

    try:
        score = int(parts[7])

        sums[district] = sums.get(district, 0) + score
        counts[district] = counts.get(district, 0) + 1

    except ValueError:
        continue

for district in sums:
    average = sums[district] / counts[district]
    print(district, "average:", round(average, 1))
```

```python
best_district = ""
best_average = -1

for district in sums:
    average = sums[district] / counts[district]

    if average > best_average:
        best_average = average
        best_district = district

print("Highest average district:", best_district)
print("Highest average:", round(best_average, 1))
```
